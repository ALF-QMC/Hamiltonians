# Kitaev-Heisenberg model

This Hamiltonian has been implemented for {cite}`PhysRevB.110.L201114,inacio2026dynamicalmagnetotropicsusceptibilitynew`. It defines a spin model on a honeycomb lattice as:

\begin{align}
\hat{H}_{\text{Spin}}&=&\sum_{\langle \boldsymbol{i},\boldsymbol{j}\rangle_\gamma}\left[K_1\hat{S}_{\boldsymbol{i}}^{\gamma} \hat{S}_{\boldsymbol{j}}^{\gamma}+\Gamma_1 \left(\hat{S}_{\boldsymbol{i}}^{\alpha} \hat{S}_{\boldsymbol{j}}^{\beta}+\hat{S}_{\boldsymbol{i}}^{\beta} \hat{S}_{\boldsymbol{j}}^{\alpha}\right)
+\Gamma'_1 \left(\hat{S}_{\boldsymbol{i}}^{\gamma} \hat{S}_{\boldsymbol{j}}^{\alpha}+\hat{S}_{\boldsymbol{i}}^{\gamma} \hat{S}_{\boldsymbol{j}}^{\beta}
+\hat{S}_{\boldsymbol{i}}^{\alpha} \hat{S}_{\boldsymbol{j}}^{\gamma}+\hat{S}_{\boldsymbol{i}}^{\beta} \hat{S}_{\boldsymbol{j}}^{\gamma}\right) \right]  \nonumber \\
& & + \sum_{\langle \boldsymbol{i},\boldsymbol{j}\rangle}  J_{1}  \hat{\boldsymbol{S}}_{\boldsymbol{i}}  \cdot \hat{\boldsymbol{S}}_{\boldsymbol{j}}.+ \sum_{\langle\langle\langle \boldsymbol{i},\boldsymbol{j}\rangle\rangle\rangle}  J_{3}  \hat{\boldsymbol{S}}_{\boldsymbol{i}}  \cdot \hat{\boldsymbol{S}}_{\boldsymbol{j}},
\end{align}

where $\langle \boldsymbol{i},\boldsymbol{j} \rangle_\gamma$ refers to nearest-neighbor bonds of type $\gamma=x,y,z$ with cyclic permutations $(\gamma,\alpha,\beta)$ of $(x,y,z)$, $\langle \boldsymbol{i},\boldsymbol{j} \rangle$ to all nearest-neighbor bonds, and $\langle\langle\langle \boldsymbol{i},\boldsymbol{j}\rangle\rangle\rangle$ to third-neighbor bonds.
A magnetic field is included through a Zeeman coupling,
\begin{equation}
\hat{H}=\hat{H}_{\text{Spin}} +\mu_B\sum_{\boldsymbol{i}} g  \hat{\boldsymbol{S}}_{\boldsymbol{i}} \cdot \boldsymbol{B}  ,
\end{equation}
with an anisotropic $g$-tensor consistent with the crystallographic axes of $\alpha{\rm -RuCl}_3$.
In $\alpha{\rm -RuCl}_3$, the orthonormal axes $\mathbf{e}_a$, and $\mathbf{e}_b$, and $\mathbf{e}_c$ are defined by the crystallographic directions, with $\mathbf{e}_c \parallel [111]$, $\mathbf{e}_a \parallel [11\bar{2}]$, and $\mathbf{e}_b \parallel [\bar{1}10]$.
In this convention, $[xyz]$ denote the cubic spin-space axes, $\mathbf{e}_a$ and $\mathbf{e}_b$ lie in the honeycomb plane, and $\mathbf{e}_c$ is perpendicular to it. 
Within the honeycomb plane, $\mathbf{e}_b$ is parallel to one bond direction of the honeycomb lattice, while $\mathbf{e}_a$ is perpendicular to it.
The magnetic field is applied in the $a$-$c$ plane, nearly parallel to the crystallographic $a$ axis, consistent with the experimental setup.

The following model-specific parameters are available:


To use the Hamiltonian, it has to be selected via its name "Kit_Heis"
```fortran
&VAR_ham_name
ham_name = "Kit_Heis"  ! Name of Hamiltonian
/
```

The following values for `VAR_lattice` choose $4\times4 = 16$ honeycomb unit cells with 32 spins.

```fortran
&VAR_lattice
L1 = 4  ! Length in direction a_1
L2 = 4  ! Length in direction a_2
/
```

The rest of the model-specific parameters are in `VAR_Kit_Heis`.

```fortran
&VAR_Kit_Heis
Ham_J = -0.1d0  ! Nearest neighbor Heisenberg coupling
Ham_J3 = 0.1d0  ! Third nearest neighbor Heisenberg coupling
Ham_alphax = -1.0d0  ! Kitaev coupling on x-bonds
Ham_alphay = -1.0d0  ! Kitaev coupling on y-bonds
Ham_alphaz = -1.0d0  ! Kitaev coupling on z-bonds
Ham_Gx = 0.5d0  ! Gamma coupling on x-bonds
Ham_Gy = 0.5d0  ! Gamma coupling on y-bonds
Ham_Gz = 0.5d0  ! Gamma coupling on z-bonds
Ham_Gx_p = 0.0d0  ! Gamma prime coupling on x-bonds
Ham_Gy_p = 0.0d0  ! Gamma prime coupling on y-bonds
Ham_Gz_p = 0.0d0  ! Gamma prime coupling on z-bonds
Hab = 0.0579d0  ! mu_B * B / K ; Magnetic field strength
Htheta = 0.45d0  ! Magnetic field angle (in units of pi)
Ham_U = 7.14d0  ! Hubbard U, for freezing out charge fluctuations
Dtau = 0.1d0  ! Imaginary time step. Ltrot=Beta/dtau
Beta = 0.5d0  ! Inverse temperature
/
```

## Usage example

A brief usage example using [pyALF](https://github.com/ALF-QMC/pyALF).

### Use provided script to download ALF and link the Hamiltonian to it

````python
!./clone_alf.sh
````

### Import necessary pyALF classes.

````python
from py_alf import ALF_source, Simulation
````

### Create ALF source object and inspect available parameters

````python
alf_src = ALF_source(alf_dir='./ALF')
````

````python
# Note the available Hamiltonian 'Kit_Heis'.
alf_src.get_ham_names()
````

````python
# Display the default parameters for the 'Kit_Heis' Hamiltonian.
alf_src.get_default_params('Kit_Heis', include_generic=False)
````

### Create and run Simulation

````python
sim = Simulation(
    alf_src, "Kit_Heis",
    {
        'L1': 4,
        'L2': 4,
        'beta': 0.5,
     },
     machine='GNU',
     mpi=True,
    )
````

````python
# On first compile, an ALF-local version of HDF5 will be compiled, 
# which takes several minutes and echos many warnings.
sim.compile()
````

````python
sim.run()
````
